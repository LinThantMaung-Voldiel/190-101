```python
current = int(input("Current floor: "))
target = int(input("Target floor: "))

if current < 1 or current > 30 or target < 1 or target > 30:
    print("Invalid floor")
elif target > current:
    n = target - current
    print(f"Going up {n} floors")
elif target < current:
    n = current - target
    print(f"Going down {n} floors")
else:
    print("You are already on this floor")
```

```python
age = int(input("Age: "))
weekend = input("Weekend (y/n): ")

if age < 0:
    print("Invalid age")

else:

    if age < 12:
        price = 60

    elif age >= 60:
        price = 80

    else:
        price = 120

    if weekend == "y":
        price += 30

    print(f"Ticket price: {price} baht")
```

```python
balance = int(input("Balance: "))
amount = int(input("Amount: "))
if amount % 100 != 0:
    print("Amount must be a multiple of 100")
elif amount > balance:
    print("Insufficient funds")
elif amount > 20000:
    print("Over daily limit")
else:
    new_balance = balance - amount
    print(f"Dispensing {amount} baht. New balance: {new_balance}")
```

```python
units = int(input("Units: "))

if units <= 150:
    total = units * 3
elif units <= 400:
    total = (150 * 3) + ((units - 150) * 4)
else:
    total = (150 * 3) + (250 * 4) + ((units - 400) * 5)

print(f"Total bill: {total} baht")
```
